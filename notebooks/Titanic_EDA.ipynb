# 🚢 Titanic Survival Prediction - Exploratory Data Analysis

## Project Overview

The Titanic Survival Prediction project aims to predict whether a passenger survived the Titanic disaster using Machine Learning.

This notebook performs Exploratory Data Analysis (EDA) on the Titanic dataset before model training.

### Objectives

- Understand the dataset
- Identify missing values
- Explore important features
- Perform data cleaning
- Encode categorical variables
- Prepare the dataset for Machine Learning

```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display all columns
pd.set_option("display.max_columns", None)
```

```python
# Load Dataset

df = pd.read_csv("../data/train.csv")

df.head()
```

## Dataset Overview

Let's inspect the dataset and understand its structure.

## Missing Values Analysis

```python
df.isnull().sum()
```

### Observations

- Age contains missing values.
- Cabin contains a large number of missing values.
- Embarked contains a few missing values.

## Data Cleaning

```python
# Fill missing Age values

df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values

df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.isnull().sum()
```

## Encode Categorical Features

```python
# Encode Gender

df["Sex"] = df["Sex"].map({
    "male":0,
    "female":1
})

# Encode Embarked

df["Embarked"] = df["Embarked"].map({
    "S":0,
    "C":1,
    "Q":2
})

df.head()
```

## Feature Selection

```python
features = [
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "Embarked"
]

target = "Survived"

X = df[features]
y = df[target]

X.head()
```
